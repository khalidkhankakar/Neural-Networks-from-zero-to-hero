```python
from abc import ABC, abstractmethod
from typing import Dict, Iterator, List, Optional, Union,Iterable

class BaseTokenzier(ABC):
    def __init__(self, special_tokens: Optional[Dict[str, int]] = None, max_length: Optional[int] = None ):
        self.special_tokens: Dict[str, int] = special_tokens
        self.max_length: Optional[int] = max_length


    @abstractmethod
    def tokenize(self, text:str) -> List[str]:
         """Split *text* into a list of token strings.

        Parameters
        ----------
        text:
            Raw input string.

        Returns
        -------
        list[str]
            Human-readable token pieces (e.g. ``["Hello", "▁world"]``).
        """

    @abstractmethod
    def encode(self, text:str)-> List[int]:
         """Convert *text* into a list of integer token IDs.

        Parameters
        ----------
        text:
            Raw input string.

        Returns
        -------
        list[int]
            Token IDs that can be fed directly to a model.
        """

    @abstractmethod
    def decode(self, token_ids:List[int]) -> str:
        """Reconstruct a string from a sequence of token IDs.

        Parameters
        ----------
        token_ids:
            Sequence of integer token IDs.

        Returns
        -------
        str
            Decoded text.
        """

    @abstractmethod
    def get_vocab(self) -> Dict[str, int]:
        """Return the full vocabulary as ``{token_str: token_id}``."""
```

```python
class CharacterTokenizer(BaseTokenzier):
    def __init__(self, vocab:str, special_tokens:Optional[Iterable[str]], max_length: Optional[int] =None, ):
        super().__init__()
        # self._vocab:Dict[str, int] = vocab
        # self._id_to_token_map:Dict[int, str] = {v : k for k, v in vocab.items()} 

        self.vocab = vocab
        self.unique_chars = sorted(list(set(''.join(vocab))))
        self.stoi = {s: i for i, s in enumerate(self.unique_chars)}
        # adding the special_tokens 
        if special_tokens is not None:
            for name in special_tokens:
                if name not in self.stoi:
                    self.stoi[name] = len(self.stoi)
        self.itos = {i : s for s, i in self.stoi.items()}

    def tokenize(self, text:str) -> List[str]:
        list_of_str = []
        for ch in text:
            if ch not in self.stoi:
                list_of_str.append('<unk>')
            else:
                list_of_str.append(ch)
        return list_of_str


    def encode(self, text:str)-> List[int]:
        arr = []
        for i in text:
            if( self.stoi.get(i) != None):
                arr.append(self.stoi[i])
            else:
                arr.append(self.stoi['<unk>'])
        
        return arr

    def decode(self, token_ids:List[int]) -> str:
        return ''.join([self.itos[s] for s in token_ids])

    def get_vocab(self, text:str) -> Dict[str, int]:
        pass






ct = CharacterTokenizer(vocab=vocab, special_tokens=["<unk>", "<endoftext>"])
# ct.encode('name &'), ct.decode(ct.encode('name &'))
ct.tokenize('name&&')
ct.encode('name&&')
ct.decode([49, 36, 48, 40, 64, 64])
```
